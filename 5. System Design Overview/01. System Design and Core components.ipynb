# What is System Design?

System Design means **planning and designing how a software system/application will work at scale in real-world production**.

In interviews, System Design is not about writing code.
It is about answering questions like:

* How will millions of users use the system?
* How will servers handle heavy traffic?
* How will data be stored?
* How will the system remain fast, scalable, and reliable?
* What trade-offs are we making?

---

# Simple Interview Definition

> “System Design is the process of designing scalable, reliable, maintainable, and efficient software systems by choosing the right architecture, components, databases, communication patterns, and infrastructure.”

---

# Mindset Behind System Design

Interviewers mainly check **how you think**, not whether your architecture is perfect.

They want to see:

## 1. Can you break a large problem into components?

Example:
If interviewer asks:

> “Design WhatsApp”

You should think:

* Users
* Messaging service
* Online/offline status
* Database
* Media storage
* Notifications
* Scalability
* Reliability

Instead of directly jumping into coding.

---

## 2. Can you think at scale?

Example:
A simple app works for 100 users.

But what about:

* 1 million users?
* 100 million requests/day?

Then problems appear:

* Server crashes
* Slow database queries
* High latency
* Data inconsistency

System Design is solving these problems.

---

## 3. Can you identify trade-offs?

Every decision has pros and cons.

Example:

| Choice        | Advantage           | Disadvantage          |
| ------------- | ------------------- | --------------------- |
| SQL DB        | Strong consistency  | Harder to scale       |
| NoSQL DB      | Highly scalable     | Eventual consistency  |
| Monolith      | Simple initially    | Hard to scale later   |
| Microservices | Independent scaling | Complex communication |

Interviewers LOVE trade-offs.

# 1. Client Applications

## What are clients?

Clients are applications used by users.

Examples:

* Mobile app
* Web browser
* Desktop app
* Smart TV app

Examples:

* Instagram Android App
* Chrome browser opening YouTube
* Uber iOS app

---

## What does client do?

Client:

* Takes user input
* Sends requests to backend
* Displays response to users

Example:
You open Instagram → scroll feed.

Client sends request:

```http
GET /feed
```

Backend sends posts.

Client displays UI.

---

## Interview Understanding

Interviewers expect you to know:

* Clients should be lightweight
* Heavy processing should happen on backend
* APIs connect client and backend
* Clients may cache data for speed

---

## Important Concepts

### Stateless Communication

Each request should contain enough information.

Example:

```http
Authorization: Bearer token
```

---

### API Calls

Mostly:

* REST APIs
* GraphQL
* WebSockets (real-time apps)

---

## Real Interview Example

For WhatsApp:

* Mobile app is client
* Sends messages
* Maintains websocket connection

# 2. Backend Servers

## What are backend servers?

Backend servers contain:

* Business logic
* Authentication
* Validation
* Processing
* API handling

They are the “brain” of the system.

---

## Example

When user logs in:

Client → Backend:

```json
{
  "email": "abc@gmail.com",
  "password": "123"
}
```

Backend:

* Validates password
* Generates JWT token
* Fetches user data
* Returns response

---

# Responsibilities of Backend

## API Handling

Example:

```http
POST /login
GET /feed
POST /payment
```

---

## Business Logic

Example:
In Uber:

* Matching driver
* Fare calculation
* Surge pricing

---

## Authentication & Authorization

* JWT
* OAuth
* Sessions

---

## Data Processing

* Image compression
* Payment processing
* Recommendation engine

---

# Interview Important Concepts

---

## Stateless Servers

Modern backend servers are usually stateless.

Meaning:

* Any server can handle any request
* Session not stored inside server memory

This improves scaling.

---

## Horizontal Scaling

Instead of:

* One powerful server

Use:

* Many smaller servers

Example:

```text
Users → Load Balancer → 10 Backend Servers
```

---

## Why multiple servers?

Because:

* Traffic increases
* Fault tolerance improves
* High availability achieved

# Interview Important Concepts

---

# Replication

Copying data to multiple servers.

Purpose:

* High availability
* Fault tolerance

Example:
Primary DB → Replica DBs

Reads from replicas.

---

# Sharding

Splitting data across multiple databases.

Example:

* Users A-M → DB1
* Users N-Z → DB2

Used for scaling huge systems.

---

# Caching

Frequently used data stored temporarily.

Example:
Redis cache.

Instead of:

```text
Request → Database
```

Do:

```text
Request → Cache → Database
```

Faster response.

---

# CAP Theorem (Very Important)

Distributed systems can guarantee only two:

* Consistency
* Availability
* Partition Tolerance

Trade-offs matter.

# 4. Load Balancers

## What is Load Balancer?

Load balancer distributes traffic across servers.

Without it:

* One server overloaded
* Other servers idle

---

# Flow

```text
Users
   ↓
Load Balancer
   ↓
Server1
Server2
Server3
```

---

# Why Important?

## Prevents overload

Traffic distributed evenly.

---

## High Availability

If one server dies:

* Requests go to healthy servers

---

## Scalability

Can add more servers easily.

---

# Load Balancing Algorithms

---

## Round Robin

Requests distributed sequentially.

Example:
1 → Server1
2 → Server2
3 → Server3

Simple and common.

---

## Least Connections

Traffic sent to server with least active requests.

Better for uneven workloads.

---

## IP Hashing

Same user routed to same server.

Useful for sticky sessions.

---

# Interview Concepts

---

## Layer 4 vs Layer 7

### Layer 4

Works at TCP level.

Fast.

---

### Layer 7

Works at HTTP level.

Can route based on:

* URL
* Headers
* Cookies

More intelligent.

---

# Popular Load Balancers

* NGINX
* HAProxy
* AWS ELB

# 5. Message Queues

## What is Message Queue?

Queues help systems process tasks asynchronously.

Instead of:

```text
Request waits until task finishes
```

Do:

```text
Request → Queue → Worker processes later
```

---

# Why Needed?

Some tasks are slow:

* Sending emails
* Video processing
* PDF generation
* Notifications

If done synchronously:

* User waits
* System becomes slow

---

# Example

User uploads video.

Instead of:

```text
Upload → Process video immediately
```

Do:

```text
Upload → Queue job → Worker processes
```

User gets instant response.

---

# Popular Queues

* Kafka
* RabbitMQ
* AWS SQS

---

# Interview Concepts

---

## Producer

Sends message.

Example:
Backend service.

---

## Consumer

Processes message.

Example:
Worker service.

---

## Advantages

* Better scalability
* Loose coupling
* Retry mechanisms
* Fault tolerance

---

# Real-World Example

E-commerce Order System:

* Order placed
* Payment service
* Inventory service
* Email service

All communicate via queues/events.

Pros:

* Strong consistency

# Interviewers Want This Thought Process

Not:

> “Use MongoDB because modern.”

But:

> “Since the application requires massive write scalability and can tolerate eventual consistency, Cassandra/MongoDB may be a better choice.”

That sounds senior-level.

# Typical System Design Flow in Interviews

When interviewer asks:

> “Design XYZ”

Follow this order:

---

## 1. Clarify Requirements

Ask:

* Daily active users?
* Read-heavy or write-heavy?
* Real-time needed?
* Consistency requirements?

---

## 2. Estimate Scale

Example:

* 10M users
* 1M requests/sec
* Storage needed

---

## 3. High-Level Design

Draw:

```text
Client
↓
Load Balancer
↓
Backend
↓
Database + Cache + Queue
```

---

## 4. Deep Dive

Discuss:

* Scaling
* Replication
* Caching
* Queues
* Fault tolerance

---

## 5. Bottlenecks

Mention:

* DB bottleneck
* Cache misses
* Network latency

---

## 6. Trade-Offs

Always discuss alternatives.

---

# Golden Rule for Interviews

* Perfect architecture

* Structured thinking
* Scalability awareness

* Trade-offs
* Real-world engineering understanding
