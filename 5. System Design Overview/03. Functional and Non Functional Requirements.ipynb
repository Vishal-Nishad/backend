# Functional vs Non-Functional Requirements

---

# 1. Functional Requirements

## Meaning

Functional requirements mean:

> “What features the system should provide.”

These are actual functionalities users use.

---

# Real Production Example — Instagram

Functional requirements:

* User can upload photo
* User can like post
* User can comment
* User can follow users

These are FEATURES.

---

# Simple Interview Explanation

If interviewer asks:

> “What are functional requirements?”

You say:

> “Functional requirements define what the system should do from the user/business perspective.”

Done.

---

# 2. Non-Functional Requirements

## Meaning

Non-functional requirements mean:

> “How well the system should work.”

Not features.

But qualities of system.

---

# Real Production Example — Instagram

Non-functional requirements:

* App should not crash with 10M users
* Feed should load fast
* System should be secure
* App should remain available

These are QUALITY attributes.

---

# Simple Interview Explanation

> “Non-functional requirements define system qualities like scalability, performance, reliability, and security.”

---

# Easy Difference

| Functional       | Non-Functional        |
| ---------------- | --------------------- |
| WHAT system does | HOW WELL system works |
| Features         | Quality               |

---

# Example Together

## WhatsApp

### Functional

* Send messages
* Create groups
* Share images

### Non-Functional

* Messages should deliver quickly
* System should support millions of users
* Chats should be secure

---

# Now Important Non-Functional Requirements

Only understand practical meaning.

---

# 1. Scalability

## Meaning

Can system handle increasing users/load?

---

# Real Example

Your backend works fine for:

* 1,000 users

But crashes at:

* 1 million users

Then:

* System is NOT scalable

---

# Production Example

Instagram:

* Millions upload posts simultaneously

To handle this:

* Multiple backend servers
* Load balancers
* Caching

---

# Interview-Friendly Line

> “Scalability means the system can handle increasing traffic without major performance degradation.”

---

# 2. Availability

## Meaning

Is the system UP and accessible?

---

# Real Example

Suppose:

* Swiggy servers go down during dinner time

Users cannot order food.

Low availability.

---

# Production Goal

Systems aim for:

* 99.9%
* 99.99% uptime

---

# Simple Interview Line

> “Availability means the system remains operational and accessible when users need it.”

---

# 3. Reliability

## Meaning

Does system work correctly consistently?

---

# Real Example

Suppose:

* Payment deducted twice
* Message lost
* Wrong bank balance shown

System is unreliable.

Even if server is UP.

---

# Important Difference

| Availability      | Reliability    |
| ----------------- | -------------- |
| System accessible | System correct |

---

# Real Production Example

UPI system:

* Must reliably process transactions

Wrong money transfer = disaster.

---

# Interview Line

> “Reliability means the system consistently performs correct operations without failures or incorrect behavior.”

---

# 4. Performance

## Meaning

How FAST system responds.

---

# Real Example

You open Instagram feed:

* Loads in 100ms → good performance
* Loads in 10 sec → bad performance

---

# Production Focus

Performance means:

* Fast APIs
* Fast DB queries
* Low latency

---

# Common Optimizations

* Redis cache
* CDN
* DB indexing

---

# Interview Line

> “Performance measures how quickly and efficiently the system responds to requests.”

---

# 5. Security

## Meaning

Protecting:

* User data
* APIs
* Systems

from unauthorized access.

---

# Real Example

If passwords stored in plain text:

* Huge security problem

---

# Production Security Examples

* JWT authentication
* HTTPS
* Password hashing
* Rate limiting

---

# Interview Line

> “Security ensures only authorized users can access resources while protecting sensitive data.”

---

# 6. Maintainability

## Meaning

How easy system is to:

* Debug
* Modify
* Add features
* Fix bugs

---

# Real Example

Bad codebase:

* Small change breaks entire app

Good maintainability:

* Easy feature addition

---

# Production Example

Microservices help teams work independently.

---

# Interview Line

> “Maintainability means the system is easy to understand, modify, test, and extend over time.”

---

# 7. Observability

This is very important in real jobs.

---

# Meaning

Can we understand:

* What is happening inside production system?

---

# Real Example

Suppose:

* API suddenly becomes slow

How will engineers find issue?

Using:

* Logs
* Monitoring
* Metrics

---

# Production Tools

* Grafana
* Prometheus
* ELK

---

# Real Example

You deployed backend.

Suddenly:

* Error rate increased

Observability helps detect:

* Failed DB queries
* Memory leak
* Slow APIs

---

# Interview Line

> “Observability helps engineers monitor, debug, and understand production system behavior using logs, metrics, and tracing.”

---

# Most Important Thing For YOU (1+ YOE)

You do NOT need:

* Deep CAP theorem initially
* Consensus algorithms
* Complex distributed theory

Instead focus on:

* Why systems fail
* Why scaling needed
* Why cache used
* Why queues used
* Why monitoring important

Because these are real production engineering concepts.

---

# Best Way To Learn System Design Initially

For every concept ask:

### 1. What problem does it solve?

### 2. Why needed in production?

### 3. What happens without it?

### 4. Real company example?

That is enough initially.

# Example Thinking

## Redis Cache

### Problem:

Database becomes slow under heavy traffic.

### Solution:

Store frequently accessed data in Redis.

### Real Example:

Instagram feed caching.

That’s how you should learn.
